```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

## Task 1 — Scatter: life expectancy change over time

**What to build:** A scatter showing **GDP per capita vs life expectancy** for **two years** (2002 and 2007) to show how both moved — use **colour for year** (just 2 colours), **one continent only**.

Choose any continent except Africa (that was the example). Highlight the change direction.

> 💡 Filter: `df.loc[df['continent'] == 'YOUR_CHOICE']` then filter years


```python
# Task 1
# YOUR CODE HERE
import plotly.express as px

import plotly.graph_objects as go

# Task 1: Asia, 2002 vs 2007

asia = df.loc[(df['continent'] == 'Asia') & (df['year'].isin([2002, 2007]))].copy()

asia['year'] = asia['year'].astype(str)

fig = px.scatter(

    asia, x='gdpPercap', y='lifeExp', color='year',

    log_x=True, hover_name='country',

    color_discrete_map={'2002': '#B0B0B0', '2007': '#1F77B4'},

    labels={'gdpPercap': 'GDP per capita (US$, log scale)',

            'lifeExp': 'Life expectancy (years)', 'year': 'Year'},

)

# thin grey lines show each country's move from 2002 to 2007

for country, g in asia.groupby('country'):

    g = g.sort_values('year')

    fig.add_trace(go.Scatter(

        x=g['gdpPercap'], y=g['lifeExp'], mode='lines',

        line=dict(color='#D9D9D9', width=1),

        showlegend=False, hoverinfo='skip'))

fig.update_traces(marker=dict(size=9), selector=dict(mode='markers'))

fig.update_layout(

    title='Between 2002 and 2007, almost every Asian country got richer and lived longer',

    template='simple_white', legend_title_text='Year', width=850, height=550)

fig.show()
```

Task 2 — Bubble chart: tell a story
What to build: A bubble chart (full 2007 dataset, all countries) where:

x = GDP per capita (log scale)
y = life expectancy
size = population
colour = ONE continent highlighted (your choice), all others grey
At least one annotation explaining the highlighted group's story
This is the grey-and-highlight technique applied to a bubble chart.

```python
# Task 2
# YOUR CODE HERE
import plotly.express as px

# 1. Select African countries for 2002 and 2007

africa = df[

    (df["continent"] == "Africa") &

    (df["year"].isin([2002, 2007]))

].copy()

# 2. Create the bubble chart

fig = px.scatter(

    africa,

    x="gdpPercap",

    y="lifeExp",

    size="pop",

    color="year",

    hover_name="country",

    hover_data={

        "country": False,

        "year": True,

        "gdpPercap": ":,.0f",

        "lifeExp": ":.1f",

        "pop": ":,.0f"

    },

    size_max=35,

    opacity=0.7,

    log_x=True,

    color_discrete_map={

        2002: "#2878B5",

        2007: "#E76F51"

    },

    labels={

        "gdpPercap": "GDP per Capita (USD, log scale)",

        "lifeExp": "Life Expectancy (years)",

        "pop": "Population",

        "year": "Year"

    },

    title="Africa (2002–2007): Comparing Wealth, Health and Population",

    template="plotly_white"

)

# 3. Add an explanatory annotation

fig.add_annotation(

    x=1500,

    y=50,

    text="African countries show<br>differences in income<br>and life expectancy",

    showarrow=True,

    arrowhead=2,

    ax=70,

    ay=-50,

    bgcolor="white",

    bordercolor="#E76F51",

    borderwidth=1

)

# 4. Improve layout and readability

fig.update_layout(

    height=650,

    width=1050,

    legend_title_text="Year",

    margin=dict(l=80, r=50, t=100, b=80),

    font=dict(size=12)

)

fig.update_xaxes(showgrid=True, gridwidth=0.5)

fig.update_yaxes(showgrid=True, gridwidth=0.5)

fig.show()
```
